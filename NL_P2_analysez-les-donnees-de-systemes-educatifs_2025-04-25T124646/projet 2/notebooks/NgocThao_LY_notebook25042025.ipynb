```python
year_columns = [col for col in EdStatsData_with_country_income_serie_def_cleaned.columns if col.isdigit()]

#colonne "nb_valeurs" qui compte le nombre de valeurs non nulles pour chaque ligne (indicateur/pays)
EdStatsData_with_country_income_serie_def_cleaned['nb_valeurs'] = EdStatsData_with_country_income_serie_def_cleaned[year_columns].notnull().sum(axis=1)
```

```python
EdStatsData_with_country_income_serie_def_cleaned
```

```python
grouped = EdStatsData_with_country_income_serie_def_cleaned[EdStatsData_with_country_income_serie_def_cleaned['nb_valeurs'] > 0].groupby(['Indicator Name_x', 'Country Name'])

nb_pays_par_indicateur = (
    grouped.size()
    .reset_index(name='count')
    .groupby('Indicator Name_x')
    .size()
    .reset_index(name='nb_pays_avec_donnees')
    .sort_values(by='nb_pays_avec_donnees', ascending=False)
)

print(nb_pays_par_indicateur.head(10))
```

```python

nb_valeurs_par_pays = (
    EdStatsData_with_country_income_serie_def_cleaned.groupby('Country Name')['nb_valeurs']
    .sum()
    .reset_index()
    .sort_values(by='nb_valeurs', ascending=False)
)

print(nb_valeurs_par_pays.head(10))
```

```python
nb_valeurs_par_pays.set_index('Country Name')['nb_valeurs']
```

```python
EdStatsData_with_country_income_serie_def_cleaned['total_valeurs_par_pays'] = (EdStatsData_with_country_income_serie_def_cleaned['Country Name'].map(nb_valeurs_par_pays.set_index('Country Name')['nb_valeurs'])
)

EdStatsData_sorted = EdStatsData_with_country_income_serie_def_cleaned.sort_values(by='total_valeurs_par_pays', ascending=False)

EdStatsData_sorted = EdStatsData_sorted.drop(columns='total_valeurs_par_pays')
EdStatsData_sorted
```
